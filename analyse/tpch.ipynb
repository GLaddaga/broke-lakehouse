# TPC-H, read from `gold`

The sample a researcher starts from. Needs only `broke_lakehouse` and a reader's Infisical identity in the environment (@docs/access.md).

```python
from broke_lakehouse.client import LAKE, attach

con = attach("prod")
```

What there is to read:

```python
con.sql(f"""
    SELECT schema_name, table_name
    FROM duckdb_tables()
    WHERE database_name = '{LAKE}'
    ORDER BY ALL""")
```

Revenue by region, all time:

```python
con.sql(f"""
    SELECT region, sum(net_revenue) AS net_revenue, sum(orders) AS orders
    FROM {LAKE}.gold.revenue_monthly__tpch
    GROUP BY region
    ORDER BY net_revenue DESC""")
```

One region, month by month and segment:

```python
con.sql(f"""
    SELECT month, market_segment, net_revenue, customers
    FROM {LAKE}.gold.revenue_monthly__tpch
    WHERE region = 'EUROPE'
    ORDER BY month, market_segment
    LIMIT 25""")
```
